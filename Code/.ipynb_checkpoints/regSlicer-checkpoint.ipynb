```python
import slicer
import numpy as np
import ast
import os

i = 0  # or your desired index

# Load CT and MR images
ctNode = slicer.util.loadVolume(os.path.join('../Data/proc/', reg_df.nii_filepath[i]), returnNode=True)[1]
mrNode = slicer.util.loadVolume(os.path.join('../Data/proc/', reg_df.nii_filepath[i+1]), returnNode=True)[1]

# # Create a transform node and set the 4x4 matrix
matrix = np.array(ast.literal_eval(reg_df.TransformationMatrix[i+1])).reshape(4, 4)
transformNode = slicer.mrmlScene.AddNewNodeByClass('vtkMRMLLinearTransformNode')
transformNode.SetMatrixTransformToParent(slicer.util.vtkMatrixFromArray(matrix))

# Save the transform as a .h5 file (BRAINSResample expects a file)
transform_path = os.path.join(slicer.app.temporaryPath, "temp_transform.h5")
slicer.util.saveNode(transformNode, transform_path)

# Create a new output node for the resampled MR image
resampledMrNode = slicer.mrmlScene.AddNewNodeByClass('vtkMRMLScalarVolumeNode')

print(slicer.mrmlScene.IsNodePresent(resampledMrNode))  # Should be 1

# Set up BRAINSResample parameters
parameters = {
    'inputVolume': mrNode.GetID(),
    'referenceVolume': ctNode.GetID(),
    'outputVolume': resampledMrNode.GetID(),
    'warpTransform': transform_path,
    'interpolationMode': 'Linear'
}

# parameters = {
#     'inputVolume': mrNode.GetID(),
#     'referenceVolume': ctNode.GetID(),
#     'outputVolume': resampledMrNode.GetID(),
#     'interpolationMode': 'Linear'
# }

cliNode = slicer.cli.runSync(slicer.modules.brainsresample, None, parameters)

# Always get the output node by ID after the CLI runs
outputNodeID = parameters['outputVolume']
outputNode = slicer.mrmlScene.GetNodeByID(outputNodeID)
print("Output node present:", slicer.mrmlScene.IsNodePresent(outputNode))  # Should be 1
print("Output node type:", outputNode.GetClassName())
print("Output node valid:", outputNode.GetImageData() is not None)

# Save as NRRD
output_path_nrrd = '/Users/caryngeady/Desktop/out/transformedBRAINS.nrrd'
success = slicer.util.saveNode(outputNode, output_path_nrrd)
print("NRRD save success:", success)
```
